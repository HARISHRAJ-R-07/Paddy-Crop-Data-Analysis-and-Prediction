
# Model Building

## Regression Models

Regression is a supervised machine learning technique used to predict continuous numerical values. In this project, the objective is to predict paddy yield using relevant agricultural features.

The following regression algorithms will be compared:

- **Linear Regression:** A baseline model that estimates a linear relationship between input features and paddy yield.
- **Decision Tree Regressor:** Learns decision rules from the data to predict paddy yield.
- **Random Forest Regressor:** Combines multiple decision trees to improve prediction performance and reduce overfitting.
- **Gradient Boosting Regressor:** Builds trees sequentially to improve prediction errors made by earlier trees.

##Cross-Validation

Cross-validation is used to evaluate model performance across different subsets of the training data.

In this project, 5-fold cross-validation will be used. The training dataset is divided into five folds. In each iteration, four folds are used for training and the remaining fold is used for validation.

This process is repeated five times, and the average evaluation metrics are used to compare model performance.

### Cross-Validation Metrics

#### Mainly,

- **MAE (Mean Absolute Error):** Measures the average absolute difference between actual and predicted yield. Lower values indicate better performance.
- **RMSE (Root Mean Squared Error):** Measures prediction errors while giving greater importance to larger errors. Lower values indicate better performance.
- **R² Score:** Measures how well the model explains the variation in paddy yield. A higher value generally indicates better performance.

## Hyperparameter Tuning

Hyperparameter tuning is the process of finding suitable model settings to improve prediction performance.

In this project, RandomizedSearchCV will be used to test different hyperparameter combinations using 5-fold cross-validation.

Examples of hyperparameters include:

- **Decision Tree:** Maximum depth and minimum samples per leaf.
- **Random Forest:** Number of trees, maximum depth, and minimum samples per leaf.
- **Gradient Boosting:** Learning rate, number of trees, and tree depth.

The model with the lowest cross-validation RMSE will be selected as the best candidate and evaluated on an unseen test dataset.

## Model Comparison and Selection

All candidate regression models will be compared using cross-validation results and test-set evaluation metrics.

The final model will be selected based on prediction error, consistency across validation folds, and generalization performance on unseen data.

The selected model and its evaluation results will be documented after the experiments are completed.


````python
# Import required packages for modeling

from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.metrics import (
    accuracy_score,
    recall_score,
    f1_score, confusion_matrix,
    classification_report,
    precision_score,
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
````

#`Cross Validation`

````python
# 5-Fold Cross-Validation

from sklearn.model_selection import KFold, GridSearchCV, RandomizedSearchCV
cv_state = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
````

````python
from scipy.stats import randint, uniform

from sklearn.pipeline import Pipeline # Pipeline connects multiple machine-learning steps together and runs them in the correct order automatically

# Parameters for using models
# Decision Tree parameters
dt_params = {
    "model__max_depth": [None, 5, 10, 15, 20],
    "model__min_samples_split": randint(2, 15),
    "model__min_samples_leaf": randint(1, 8),
    "model__max_features": [None, "sqrt", 1.0]
}

# Random Forest parameters
rf_params = {
    "model__n_estimators": randint(100, 301),
    "model__max_depth": [None, 5, 10, 15, 20],
    "model__min_samples_split": randint(2, 15),
    "model__min_samples_leaf": randint(1, 8),
    "model__max_features": [1.0, "sqrt", "log2"]
}

# Gradient Boosting parameters
gb_params = {
    "model__n_estimators": randint(50, 251),
    "model__learning_rate": uniform(0.01, 0.19),
    "model__max_depth": randint(2, 6),
    "model__min_samples_split": randint(2, 15),
    "model__min_samples_leaf": randint(1, 8),
    "model__subsample": uniform(0.7, 0.3)
}
````

````python
# Create regression models

models = {
    "Decision Tree": (
        DecisionTreeRegressor(random_state=42),
        dt_params
    ),

    "Random Forest": (
        RandomForestRegressor(
            random_state=42,
            n_jobs=1
        ),
        rf_params
    ),

    "Gradient Boosting": (
        GradientBoostingRegressor(random_state=42),
        gb_params
    )
}

# Store tuning results
tuning_results = []
best_estimators = {}
````

````python
# Hyperparameter tuning using RandomizedSearchCV

from sklearn.impute import SimpleImputer

for model_name, (regressor, params) in models.items():

    print(f"\nTuning {model_name}...")

    model_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", regressor)
    ])

    random_search = RandomizedSearchCV(
        estimator=model_pipeline,
        param_distributions=params,
        n_iter=20,
        scoring="neg_root_mean_squared_error",
        cv=cv_state,
        random_state=42,
        n_jobs=-1,
        refit=True,
        return_train_score=True
    )

    random_search.fit(x_train, y_train)

    best_estimators[model_name] = random_search.best_estimator_

    tuning_results.append({
        "Model": model_name,
        "Best CV RMSE": -random_search.best_score_,
        "Best Parameters": random_search.best_params_
    })

    print("Best Parameters:", random_search.best_params_)
    print("Best CV RMSE:", round(-random_search.best_score_, 3))
````

````python
# Compare cross-validation results

tuning_results_df = pd.DataFrame(tuning_results)

tuning_results_df = tuning_results_df.sort_values(
    by="Best CV RMSE",
    ascending=True
).reset_index(drop=True)

tuning_results_df
````

````python
# Select the best model using cross-validation

best_model_name = tuning_results_df.loc[0, "Model"]
best_model = best_estimators[best_model_name]

print("Best Model:", best_model_name)

# Predict on unseen test data
y_pred = best_model.predict(x_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\nFinal Test Results")
print("------------------")
print("MAE:", round(mae, 3))
print("RMSE:", round(rmse, 3))
print("R2 Score:", round(r2, 4))
````

````python
# Compare all tuned models on the same test set

test_results = []

for model_name, model in best_estimators.items():

    predictions = model.predict(x_test)

    test_results.append({
        "Model": model_name,
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(
            mean_squared_error(y_test, predictions)
        ),
        "R2 Score": r2_score(y_test, predictions),
    })

test_results_df = pd.DataFrame(test_results).sort_values(
    by="RMSE",
    ascending=True
).reset_index(drop=True)

test_results_df
````

### 2. Observations of Overfitting or Underfitting

Overfitting occurs when a model performs very well on the training data but performs noticeably worse on unseen test data. Underfitting occurs when a model performs poorly on both training and test data.

To identify these patterns, the training and test RMSE and R² scores should be compared. The cross-validation RMSE values alone are not sufficient to confirm whether a model is overfitting or underfitting.

### 3. Bias–Variance Considerations

A high-bias model may be too simple to capture important relationships in the dataset, resulting in underfitting. A high-variance model may learn noise or specific patterns from the training data, resulting in overfitting.

Cross-validation and hyperparameter tuning help balance bias and variance and improve the model's ability to generalize to unseen paddy yield data. In this evaluation, the Decision Tree Regressor achieved the lowest cross-validation RMSE, although its generalization performance must also be assessed using the test dataset.

### Conclusion
